```python
import pandas as pd

df = pd.read_csv('insurance.csv')
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
import matplotlib.pyplot as plt

df['age'].hist(bins=20)
plt.title('Распределение возраста')
plt.xlabel('Возраст')
plt.ylabel('Количество')
plt.show()
```

```python
df.groupby('smoker')['charges'].mean().plot(kind='bar')
plt.title('Средние расходы: курящие vs некурящие')
plt.ylabel('Расходы')
plt.show()
```

```python
df['region'].value_counts().plot(kind='bar')
plt.title('Количество по регионам')
plt.show()
```
